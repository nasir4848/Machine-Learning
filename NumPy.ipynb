**NumPy Practice**

```python
import numpy as np
numbers=np.array([10,20,30,40,50])
print(numbers)
```

**Accessing Array Elements**

```python
num=np.array([3,6,9,15])
print(num[0])
print(num[1])
print(num[2])
print(num[3])
print(num[-1])
```

**Mathematical Operations using NumPy**

```python
 a = np.array([10,15,20,25,30])
 print(a+10)
 print(a-5)
 print(a*5)
 print(a/2)
 print(a%3)
```

**Important NumPy Functions**

```python
mark=np.array([20,30,40,50,60])
print(np.sum(mark))
print(np.mean(mark))
print(np.median(mark))
print(np.max(mark))
print(np.min(mark))
print(np.sort(mark))
print(np.sqrt(mark))
print(np.square(mark))
print(np.cos(mark))
```

**list Shape Change**

```python
li=np.array([10,20,30,40,50,60])
print(li.shape)

new_li=li.reshape(2,3)
print(new_li)
```

**Slicing of Data**

```python
b=np.array([10,20,30,40,50,60])

print(b[0:])
print(b[1:5])
print(b[:3])
print(b[2:])
print(b[::-1])
```

**Filtering Data using Condition**

```python
marks=np.array([25,50,60,80,30,90])

print(marks[marks>=50])
print(marks[marks<50])
print(marks[marks==60])
```

**Uses of np.where() using condition**

```python
res=np.array([20,40,50,60,70,30])

result=np.where(res>=40, "Pass","Fail")
print(result)
```
