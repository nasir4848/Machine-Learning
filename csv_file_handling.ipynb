CSV = Comma Separated Values || Creating csv file

```python
import pandas as pd

data={
    "Name":["Nasir","Rahim","Karim","Alice"],
    "Age":[21,22,23,24],
    "Marks":[70,80,75,90]
}

df=pd.DataFrame(data)
df.to_csv("students.csv",index=False)
print("CSV file created Successfull")
```
