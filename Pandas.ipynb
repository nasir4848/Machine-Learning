**Creating a DataFrame**

```python
data={
    "Name":["Nasir","Rahim","Karim"],
    "Age":[21,22,23],
    "Marks":[80,75,85]
}
df=pd.DataFrame(data)
print(df)
```

**Selecting Columns**

```python
info={
    "Name":["Nasir","Rahim","Karim"],
    "Age":[21,22,23],
    "Marks":[80,70,85]
}
df=pd.DataFrame(info)

print(df["Marks"])
print(df["Age"])
print(df[["Name", "Age"]])   #for print two columns
```

**Filtering Data**

```python
data={
    "Name":["Nasir","Rahim","Karim","Alice","Johnson"],
    "Marks":[90,70,45,85,50]
}

df=pd.DataFrame(data)
print(df[df["Marks"]<=70])     #Direct print using print function
print("\n")
result=df[df["Marks"]>=70]      #First store in result variable and then print
print(result)
```

**Important Pandas Functions**

```python
data={
    "Name":["Nasir","Tahin","Rahim","karim","Alice","Johnson","Orion"],
    "Marks":[40,50,20,70,80,90,65]
}

df=pd.DataFrame(data)
print(df.head())
print(df.head(3))

print(df.tail())
print(df.tail(3))

print(df.shape)
print(df.columns)

print(df["Marks"].mean())
print(df["Marks"].max())
print(df["Marks"].min())
```

**uses of Groupby() **

```python
data={
    "Dept":["CSE","EEE","CSE","EEE"],
    "Name":["a","b","c","d"],
    "Marks":[80,30,40,50]
}

df=pd.DataFrame(data)
print(df.groupby("Dept")["Marks"].mean())
```

**Value Counts**

```python
data={
    "Dept":["CSE","EEE","CSE","EEE","CE"],
    "Name":["a","b","c","d","e"],
    "Marks":[80,30,40,50,70]
}

df=pd.DataFrame(data)
print(df["Dept"].value_counts())
```

**Searching Duplicates**

```python
data={
    "Dept":["CSE","EEE","CSE","EEE","CE"],

    "Marks":[80,30,40,50,70]
}

df=pd.DataFrame(data)
print(df.duplicated(subset="Dept"))

df=df.drop_duplicates(subset="Dept")
print(df)
```
